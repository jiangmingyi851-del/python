````python
def calculate_total(Courses, Selected):
    se=set(Selected)
    total = 0
    for name, credit in Courses:
        if name in se:
            total += credit
    return total

Courses = [('三宝', 2), ('音数', 2), ('地概', 2), ('高数', 4), ('线代', 3), ('棒垒', 1)]
Selected = ['三宝', '高数', '线代']
assert calculate_total(Courses, Selected) == 9
````

````python
import random
random.seed(42)
l=[random.randint(0,10) for _ in range(100)]
def output(l):
    # TODO
    x=iter(l)
    while True:
        try:
            num=next(x)
            if num==1:
                return next(x) , next(x)
        except StopIteration:
            break

assert output(l) == (0, 4)
````

````python
# 进阶示例：文本处理函数与列表生成式
import re
from typing import List, Tuple
from collections import Counter

def clean_text(text: str) -> str:
    """清理文本：移除特殊字符、转换为小写"""
    # 移除HTML标签
    text = re.sub(r'<.*?>', '', text)
    # 正则表达式，？表示非贪婪匹配，.*表示匹配前面的元素零次或多次
    # 移除特殊字符，保留字母、数字和基本标点
    text = re.sub(r'[^\w\s.,!?-]', '', text)
    # 正则表达式，[]表示匹配字符集，^表示取反，-表示横杠，\w表示字母数字下划线，\s表示空白字符，.表示句号，,表示逗号，!表示感叹号，?表示问号，-表示连字符
    # 转换为小写
    return text.lower().strip()

def extract_emails(text: str) -> List[str]:
    """从文本中提取所有邮箱地址"""
    email_pattern = r'\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Z|a-z]{2,}\b'
    # 正则表达式，\b表示单词边界，[A-Za-z0-9._%+-]+表示至少一个字母数字下划线%+-的邮箱用户名，@[A-Za-z0-9.-]+表示@后至少一个字母数字.-的域名，\.表示句号，[A-Z|a-z]{2,}表示至少两个字母的TLD.
    return re.findall(email_pattern, text)

def calculate_word_frequencies(texts: List[str], 
                             stopwords: List[str] = None) -> List[Tuple[str, int]]:
    """计算词频，排除停用词"""
    if stopwords is None:
        stopwords = ['the', 'and', 'is', 'in', 'to', 'of', 'a', 'for']
    
    # 清理所有文本并分割单词
    all_words = [
        word 
        for text in texts 
        for word in clean_text(text).split()
        if word not in stopwords and len(word) > 2
    ]
    
    # 计算词频并排序
    word_counts = Counter(all_words)
    return sorted(word_counts.items(), key=lambda x: x[1], reverse=True)

def ngram_generator(text: str, n: int = 2) -> List[str]:
    """生成n-gram序列"""
    words = clean_text(text).split()
    if len(words) < n:
        return []
    
    return [' '.join(words[i:i+n]) for i in range(len(words) - n + 1)]

def extract_keywords(text: str, 
                    min_length: int = 3, 
                    max_length: int = 10) -> List[str]:
    """提取可能的关键词（基于长度和内容）"""
    words = clean_text(text).split()
    
    # 使用列表生成式过滤和转换
    keywords = [
        word.capitalize() 
        for word in words 
        if min_length <= len(word) <= max_length
        and word.isalpha()  # 只包含字母
        and not word.isdigit()  # 排除纯数字
    ]
    
    # 移除重复但保持顺序
    seen = set()
    unique_keywords = []
    for word in keywords:
        if word not in seen:
            seen.add(word)
            unique_keywords.append(word)
    
    return unique_keywords

# 使用示例
if __name__ == "__main__":
    # 测试文本
    sample_texts = [
        "Hello world! Email me at john.doe@example.com or visit https://example.com",
        "The quick brown fox jumps over the lazy dog. Contact: jane@test.org",
        "Python programming is FUN! For inquiries: support@python.org"
    ]
    
    # 1. 提取所有邮箱
    all_emails = [
        email 
        for text in sample_texts 
        for email in extract_emails(text)
    ]
    print(f"提取到的邮箱: {all_emails}")
    
    # 2. 计算词频
    word_freq = calculate_word_frequencies(sample_texts)
    print(f"\n词频统计(前5): {word_freq[:5]}")
    
    # 3. 生成所有文本的2-gram
    all_bigrams = [
        bigram 
        for text in sample_texts 
        for bigram in ngram_generator(text, 2)
    ]
    print(f"\n所有2-gram(前10): {all_bigrams[:10]}")
    
    # 4. 提取关键词
    keywords_list = [
        extract_keywords(text, min_length=4, max_length=8)
        for text in sample_texts
    ]
    print(f"\n各文本关键词:")
    for i, keywords in enumerate(keywords_list, 1):
        print(f"  文本{i}: {keywords}")
    
    # 5. 复杂的文本分析管道
    analysis_results = [
        {
            'text': text[:30] + '...',  # 截断显示
            'email_count': len(extract_emails(text)),
            'word_count': len(clean_text(text).split()),
            'unique_words': len(set(clean_text(text).split())),
            'keywords': extract_keywords(text)[:5]  # 取前5个关键词
        }
        for text in sample_texts
    ]
    
    print(f"\n文本分析结果:")
    for result in analysis_results:
        print(f"  文本: {result['text']}")
        print(f"    邮箱数: {result['email_count']}, 总词数: {result['word_count']}")
        print(f"    唯一词数: {result['unique_words']}, 关键词: {result['keywords']}")
````

针对“按关键词索引-检索”这一目标，传统的基于频率（如 TF-IDF）的算法往往存在两个问题：
1.  **语义缺失**：无法理解“car”和“automobile”是相似的。
2.  **冗余性**：容易提取出含义极其相近的词（如 "fast", "faster", "fastest"），导致索引浪费。

为了设计一个“更加合理”的算法，我推荐使用 **语义嵌入（Embedding）结合 最大边界相关性（MMR, Maximal Marginal Relevance）** 的方法。

### 算法设计思路
这是一个现代工业界常用的 **KeyBERT** 变体思路。
1.  **候选词生成 (Candidate Generation)**：使用 N-gram（N元语法）提取文本中的短语（不仅是单词，还有 "Machine Learning" 这样的词组）。
2.  **语义向量化 (Vectorization)**：利用轻量级预训练模型（如 BERT 的微型版本）将“原文档”和“所有候选词”映射到高维向量空间。
3.  **余弦相似度 (Cosine Similarity)**：计算候选词与文档核心含义的距离。
4.  **MMR 排序 (Diversity Ranking)**：这是核心。我们在选择关键词时，不仅要求它与原文**相似**，还要求它与**已选出的关键词不相似**。这样能保证提取出的关键词覆盖文档的不同侧面，避免同义词堆积。

```python
import numpy as np
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.metrics.pairwise import cosine_similarity
from sentence_transformers import SentenceTransformer

class SmartKeywordExtractor:
    def __init__(self, model_name='all-MiniLM-L6-v2'):
        """
        Initialize with a lightweight, efficient transformer model.
        'all-MiniLM-L6-v2' is fast and effective for English text.
        """
        print(f"Loading model: {model_name}...")
        self.model = SentenceTransformer(model_name)
        
    def _mmr(self, doc_embedding, candidate_embeddings, candidates, top_n, diversity):
        """
        Maximal Marginal Relevance (MMR) algorithm to select keywords.
        
        Args:
            doc_embedding: Vector representation of the document.
            candidate_embeddings: Matrix of vector representations of candidates.
            candidates: List of candidate strings.
            top_n: Number of keywords to extract.
            diversity: Float between 0 and 1. 
                       0 means select most relevant (redundancy possible).
                       1 means select most diverse (relevance might drop).
        """
        # Calculate similarity between document and all candidates
        word_doc_similarity = cosine_similarity(candidate_embeddings, doc_embedding)

        # Calculate similarity between candidates themselves
        word_similarity = cosine_similarity(candidate_embeddings)

        # Initialize
        keywords_idx = [np.argmax(word_doc_similarity)]
        candidates_idx = [i for i in range(len(candidates)) if i != keywords_idx[0]]

        # Iteratively select the best next keyword
        for _ in range(top_n - 1):
            candidate_similarities = word_doc_similarity[candidates_idx, :]
            target_similarities = np.max(word_similarity[candidates_idx][:, keywords_idx], axis=1)

            # MMR equation: Score = (1-diversity) * Relevance - (diversity) * Redundancy
            mmr = (1 - diversity) * candidate_similarities - diversity * target_similarities.reshape(-1, 1)
            
            mmr_idx = candidates_idx[np.argmax(mmr)]

            keywords_idx.append(mmr_idx)
            candidates_idx.remove(mmr_idx)

        return [candidates[idx] for idx in keywords_idx]

    def extract(self, text, top_n=5, keyphrase_ngram_range=(1, 2), diversity=0.5):
        """
        Main extraction function.
        
        Args:
            text (str): Input English text.
            top_n (int): Number of keywords to return.
            keyphrase_ngram_range (tuple): (min_n, max_n) e.g., (1, 2) allows "data" and "data science".
            diversity (float): 0.0 to 1.0. Higher value = more diverse topics.
        """
        # 1. Candidate Generation using CountVectorizer (removes stop words automatically)
        # We look for 1-gram to n-gram phrases
        count = CountVectorizer(ngram_range=keyphrase_ngram_range, stop_words="english").fit([text])
        candidates = count.get_feature_names_out()

        # Edge case: if text is too short
        if len(candidates) == 0:
            return []

        # 2. Embedding Generation
        # Embed the document and the candidates
        doc_embedding = self.model.encode([text])
        candidate_embeddings = self.model.encode(candidates)

        # 3. Apply MMR Ranking
        # If we request more keywords than candidates, return all candidates
        if len(candidates) <= top_n:
            return list(candidates)
            
        final_keywords = self._mmr(
            doc_embedding, 
            candidate_embeddings, 
            candidates, 
            top_n, 
            diversity
        )
        
        return final_keywords

# ==========================================

if __name__ == "__main__":
    # 模拟一段关于“自然语言处理”的英文文本

    large amounts of natural language data. The goal is a computer capable of

    print("----------------------------------------")
```

### 为什么这个设计更合理？

作为工程师，在做索引系统时，我们不仅仅关注词频，更关注**信息熵**和**覆盖率**：

1.  **支持短语 (Phrase Extraction)**：
    *   简单的算法会把 "Natural Language Processing" 拆成 "Natural", "Language", "Processing"。
    *   本程序通过 `keyphrase_ngram_range=(1, 2)` 设置，可以保留 "computer science" 或 "artificial intelligence" 作为一个完整的索引项，这对检索精度至关重要。

2.  **抗干扰能力 (Context Awareness)**：
    *   使用了 Transformer 模型（`all-MiniLM-L6-v2`），它理解上下文。如果文章在讲 "Apple" (公司) 和 "Stock" (股票)，它不会提取 "Fruit" (水果) 相关的语义，因为它知道这里的 "Apple" 是科技公司。

3.  **多样性控制 (MMR)**：
    *   在代码中你可以调整 `diversity` 参数（0.0 - 1.0）。
    *   如果是 **0.2**：结果可能很相似，例如 `['nlp', 'natural language', 'linguistics']`。
    *   如果是 **0.7**（推荐）：结果会强迫覆盖不同方面，例如 `['nlp', 'computer science', 'documents', 'program computers']`。这对建立索引非常有利，因为用户搜索不同侧面时都能命中该文档。

4.  **工程性能**：
    *   使用了 `MiniLM` 模型，速度非常快，适合在 CPU 上运行，不需要昂贵的 GPU 显卡。

这个方案是目前构建智能文档索引（Semantic Search Indexing）的业界标准做法之一。

````python
import numpy as np
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.metrics.pairwise import cosine_similarity
from sentence_transformers import SentenceTransformer

class SmartKeywordExtractor:
    def __init__(self, model_name='all-MiniLM-L6-v2'):
        """
        Initialize with a lightweight, efficient transformer model.
        'all-MiniLM-L6-v2' is fast and effective for English text.
        """
        print(f"Loading model: {model_name}...")
        self.model = SentenceTransformer(model_name)
        
    def _mmr(self, doc_embedding, candidate_embeddings, candidates, top_n, diversity):
        """
        Maximal Marginal Relevance (MMR) algorithm to select keywords.
        
        Args:
            doc_embedding: Vector representation of the document.
            candidate_embeddings: Matrix of vector representations of candidates.
            candidates: List of candidate strings.
            top_n: Number of keywords to extract.
            diversity: Float between 0 and 1. 
                       0 means select most relevant (redundancy possible).
                       1 means select most diverse (relevance might drop).
        """
        # Calculate similarity between document and all candidates
        word_doc_similarity = cosine_similarity(candidate_embeddings, doc_embedding)

        # Calculate similarity between candidates themselves
        word_similarity = cosine_similarity(candidate_embeddings)

        # Initialize
        keywords_idx = [np.argmax(word_doc_similarity)]
        candidates_idx = [i for i in range(len(candidates)) if i != keywords_idx[0]]

        # Iteratively select the best next keyword
        for _ in range(top_n - 1):
            candidate_similarities = word_doc_similarity[candidates_idx, :]
            target_similarities = np.max(word_similarity[candidates_idx][:, keywords_idx], axis=1)

            # MMR equation: Score = (1-diversity) * Relevance - (diversity) * Redundancy
            mmr = (1 - diversity) * candidate_similarities - diversity * target_similarities.reshape(-1, 1)
            
            mmr_idx = candidates_idx[np.argmax(mmr)]

            keywords_idx.append(mmr_idx)
            candidates_idx.remove(mmr_idx)

        return [candidates[idx] for idx in keywords_idx]

    def extract(self, text, top_n=5, keyphrase_ngram_range=(1, 2), diversity=0.5):
        """
        Main extraction function.
        
        Args:
            text (str): Input English text.
            top_n (int): Number of keywords to return.
            keyphrase_ngram_range (tuple): (min_n, max_n) e.g., (1, 2) allows "data" and "data science".
            diversity (float): 0.0 to 1.0. Higher value = more diverse topics.
        """
        # 1. Candidate Generation using CountVectorizer (removes stop words automatically)
        # We look for 1-gram to n-gram phrases
        count = CountVectorizer(ngram_range=keyphrase_ngram_range, stop_words="english").fit([text])
        candidates = count.get_feature_names_out()

        # Edge case: if text is too short
        if len(candidates) == 0:
            return []

        # 2. Embedding Generation
        # Embed the document and the candidates
        doc_embedding = self.model.encode([text])
        candidate_embeddings = self.model.encode(candidates)

        # 3. Apply MMR Ranking
        # If we request more keywords than candidates, return all candidates
        if len(candidates) <= top_n:
            return list(candidates)
            
        final_keywords = self._mmr(
            doc_embedding, 
            candidate_embeddings, 
            candidates, 
            top_n, 
            diversity
        )
        
        return final_keywords

# ==========================================
# Example Usage
# ==========================================

if __name__ == "__main__":
    # 模拟一段关于“自然语言处理”的英文文本
    text_input = """
    Natural language processing (NLP) is a subfield of linguistics, computer science, 
    and artificial intelligence concerned with the interactions between computers and 
    human language, in particular how to program computers to process and analyze 
    large amounts of natural language data. The goal is a computer capable of 
    understanding the contents of documents, including the contextual nuances of 
    the language within them. The technology can then accurately extract information 
    and insights contained in the documents as well as categorize and organize the documents themselves.
    """

    # 初始化提取器 (第一次运行会下载模型，约80MB)
    extractor = SmartKeywordExtractor()

    # 执行提取
    # 设置 diversity=0.7 意味着我们非常看重关键词的多样性
    keywords = extractor.extract(text_input, top_n=5, keyphrase_ngram_range=(1, 2), diversity=0.7)

    print("\n---------------- Result ----------------")
    print(f"Original Text Length: {len(text_input)} characters")
    print("Extracted Keywords/Keyphrases:")
    for i, kw in enumerate(keywords):
        print(f"{i+1}. {kw}")
    print("----------------------------------------")
````
