```python
import sys

def compute_lps(s):
    """计算字符串 s 的 LPS 数组（最长相等前后缀长度）"""
    n = len(s)
    lps = [0] * n
    j = 0
    for i in range(1, n):
        while j > 0 and s[i] != s[j]:
            j = lps[j - 1]
        if s[i] == s[j]:
            j += 1
        lps[i] = j
    return lps

def main():
    case_num = 1
    while True:
        line = sys.stdin.readline()
        if not line:                # 文件结束
            break
        n = line.strip()
        if n == '0':
            break
        n = int(n)
        # 读取 n 个字符，每行一个
        s = []
        for _ in range(n):
            ch = sys.stdin.readline().strip()
            if not ch:               # 防止意外读取不到
                break
            s.append(ch)

        lps = compute_lps(s)
        print(f"Test case #{case_num}")
        for i in range(1, n):        # 前缀长度从 2 开始，即 i+1 >= 2
            length = i + 1
            period = length - lps[i]
            # 如果 length 能被 period 整除且 period < length，说明存在重复周期
            if length % period == 0 and period < length:
                print(length, period)
        print()      # 每个测试用例后空一行
        case_num += 1

if __name__ == "__main__":
    main()
    
import sys

def compute_lps(s):
    """返回字符串s的前缀函数（LPS数组）"""
    n = len(s)
    lps = [0] * n
    j = 0
    for i in range(1, n):
        while j > 0 and s[i] != s[j]:
            j = lps[j - 1]
        if s[i] == s[j]:
            j += 1
        lps[i] = j
    return lps

def solve():
    case_no = 1
    out_lines = []
    while True:
        line = sys.stdin.readline()
        if not line:
            break
        n_str = line.strip()
        if n_str == '0':
            break
        n = int(n_str)
        s = sys.stdin.readline().strip()
        # 如果字符串可能包含空格？题目说第二行是字符串，一般无空格，直接strip即可

        lps = compute_lps(s)

        out_lines.append(f"Test case #{case_no}")
        # 只考虑长度 >= 2 的前缀
        for i in range(2, n + 1):
            period = i - lps[i - 1]          # 最小周期长度
            # 当且仅当 i 能被 period 整除且 period < i 时，存在循环节 (K > 1)
            if i % period == 0 and period < i:
                k = i // period
                out_lines.append(f"{i} {k}")
        out_lines.append("")                  # 每组数据后空行
        case_no += 1

    sys.stdout.write("\n".join(out_lines))

if __name__ == "__main__":
    solve()
```
