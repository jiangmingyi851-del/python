链表 vs 数组对比:
特性       | 数组 (顺序表)          | 链表
-----------|------------------------|---------------------------
底层实现   | 连续内存块             | 离散节点 + 指针
随机访问   | O(1) 支持              | O(n) 需从头遍历
插入/删除  | O(n) 涉及元素移动       | O(1) 已知位置时修改指针
空间开销   | 预分配，可能浪费        | 每个节点需额外存指针
Python对应 | list (动态数组)         | 需手动实现 Node 类

1. 单向链表 (Singly Linked List) - 基础版

```python
class ListNode:
    """LeetCode 风格的链表节点"""
    def __init__(self, x):
        self.val = x
        self.next = None


class Node:
    """基本链表节点"""
    def __init__(self, value):
        self.value = value
        self.next = None


class LinkedList:
    """单向链表 - 尾插法"""
    def __init__(self):
        self.head = None

    def insert(self, value):
        """尾部插入"""
        new_node = Node(value)
        if self.head is None:
            self.head = new_node
        else:
            current = self.head
            while current.next:
                current = current.next
            current.next = new_node

    def delete(self, value):
        """删除第一个匹配的节点"""
        if self.head is None:
            return
        if self.head.value == value:
            self.head = self.head.next
        else:
            current = self.head
            while current.next:
                if current.next.value == value:
                    current.next = current.next.next
                    break
                current = current.next

    def display(self):
        """打印链表"""
        current = self.head
        while current:
            print(current.value, end=" ")
            current = current.next
        print()
```

2. 单向链表 - 完整版 (含 head/tail/size)

```python
class LinkList:
    class Node:
        def __init__(self, data, next=None):
            self.data = data
            self.next = next

    def __init__(self):
        self.head = None
        self.tail = None
        self.size = 0

    def print(self):
        ptr = self.head
        while ptr is not None:
            if ptr != self.head:
                print(',', end='')
            print(ptr.data, end='')
            ptr = ptr.next
        print()

    def insert_after(self, p, data):
        """在节点 p 之后插入"""
        nd = LinkList.Node(data)
        if p is None:
            self.pushFront(data)
        else:
            nd.next = p.next
            p.next = nd
            if p == self.tail:
                self.tail = nd
            self.size += 1

    def delete_after(self, p):
        """删除节点 p 之后的节点"""
        if p is None or p.next is None:
            return
        if self.tail is p.next:
            self.tail = p
        p.next = p.next.next
        self.size -= 1

    def popFront(self):
        """弹出头部元素"""
        if self.head is None:
            raise Exception("Popping front from empty link list.")
        else:
            data = self.head.data
            self.head = self.head.next
            self.size -= 1
            if self.size == 0:
                self.tail = None
            return data

    def pushFront(self, data):
        """头部插入"""
        nd = LinkList.Node(data, self.head)
        self.head = nd
        if self.size == 0:
            self.tail = nd
        self.size += 1

    def pushBack(self, data):
        """尾部插入"""
        if self.size == 0:
            self.pushFront(data)
        else:
            self.insert_after(self.tail, data)

    def clear(self):
        self.head = None
        self.tail = None
        self.size = 0

    def __iter__(self):
        self.ptr = self.head
        return self

    def __next__(self):
        if self.ptr is None:
            raise StopIteration()
        else:
            data = self.ptr.data
            self.ptr = self.ptr.next
            return data
```

3. 双向链表 (Doubly Linked List)

```python
class DNode:
    def __init__(self, data):
        self.data = data
        self.next = None
        self.prev = None


class DoublyLinkedList:
    def __init__(self):
        self.head = None
        self.tail = None

    def append(self, data):
        """尾部添加"""
        new_node = DNode(data)
        if not self.head:
            self.head = new_node
            self.tail = new_node
        else:
            self.tail.next = new_node
            new_node.prev = self.tail
            self.tail = new_node

    def prepend(self, data):
        """头部添加"""
        new_node = DNode(data)
        if not self.head:
            self.head = new_node
            self.tail = new_node
        else:
            new_node.next = self.head
            self.head.prev = new_node
            self.head = new_node

    def delete(self, node):
        """删除指定节点"""
        if not self.head:
            return
        if node == self.head:
            self.head = node.next
            if self.head:
                self.head.prev = None
        elif node == self.tail:
            self.tail = node.prev
            if self.tail:
                self.tail.next = None
        else:
            node.prev.next = node.next
            node.next.prev = node.prev

    def print_list(self):
        """从头到尾打印"""
        current = self.head
        while current:
            print(current.data, end=" <-> ")
            current = current.next
        print("None")

    def print_reverse(self):
        """从尾到头打印"""
        current = self.tail
        while current:
            print(current.data, end=" <-> ")
            current = current.prev
        print("None")
```

4. 浏览器历史记录 (双向链表应用)
LeetCode M1472

```python
class BrowserHistory:
    class _Node:
        def __init__(self, url):
            self.url = url
            self.prev = None
            self.next = None

    def __init__(self, homepage):
        self.current = self._Node(homepage)

    def visit(self, url):
        new_node = self._Node(url)
        self.current.next = new_node
        new_node.prev = self.current
        self.current = new_node

    def back(self, steps):
        while steps > 0 and self.current.prev is not None:
            self.current = self.current.prev
            steps -= 1
        return self.current.url

    def forward(self, steps):
        while steps > 0 and self.current.next is not None:
            self.current = self.current.next
            steps -= 1
        return self.current.url
```

5. 循环链表 (Circular Linked List) - 仅设尾指针

```python
class CircleLinkList:
    class Node:
        def __init__(self, data, next=None):
            self.data = data
            self.next = next

    def __init__(self):
        self.tail = None
        self.size = 0

    def is_empty(self):
        return self.size == 0

    def pushFront(self, data):
        """头部插入"""
        nd = CircleLinkList.Node(data)
        if self.is_empty():
            self.tail = nd
            nd.next = self.tail  # 自己指向自己形成环
        else:
            nd.next = self.tail.next  # 新节点指向当前头节点
            self.tail.next = nd       # 当前尾节点指向新节点
        self.size += 1

    def pushBack(self, data):
        """尾部插入"""
        nd = CircleLinkList.Node(data)
        if self.is_empty():
            self.tail = nd
            nd.next = self.tail
        else:
            nd.next = self.tail.next  # 新节点指向当前头节点
            self.tail.next = nd       # 当前尾节点指向新节点
            self.tail = nd            # 更新尾指针
        self.size += 1

    def popFront(self):
        """移除并返回头部元素"""
        if self.is_empty():
            return None
        old_head = self.tail.next
        if self.size == 1:
            self.tail = None
        else:
            self.tail.next = old_head.next
        self.size -= 1
        return old_head.data

    def popBack(self):
        """移除并返回尾部元素"""
        if self.is_empty():
            return None
        elif self.size == 1:
            data = self.tail.data
            self.tail = None
            self.size -= 1
            return data
        else:
            prev = self.tail
            while prev.next != self.tail:
                prev = prev.next
            data = self.tail.data
            prev.next = self.tail.next
            self.tail = prev
            self.size -= 1
            return data

    def printList(self):
        """打印所有元素"""
        if self.is_empty():
            print('Empty!')
        else:
            ptr = self.tail.next
            while True:
                print(ptr.data, end=', ' if ptr != self.tail else '\n')
                if ptr == self.tail:
                    break
                ptr = ptr.next
```

6. 循环链表 - 带 remove 方法 (OJ题目版)

```python
class CircleLinkListWithRemove:
    class Node:
        def __init__(self, data, next=None):
            self.data, self.next = data, next

    def __init__(self):
        self.tail = None
        self.size = 0

    def pushFront(self, data):
        nd = self.Node(data)
        if self.tail is None:
            self.tail = nd
            nd.next = self.tail
        else:
            nd.next = self.tail.next
            self.tail.next = nd
        self.size += 1

    def pushBack(self, data):
        self.pushFront(data)
        self.tail = self.tail.next

    def popFront(self):
        if self.size == 0:
            return None
        nd = self.tail.next
        self.size -= 1
        if self.size == 0:
            self.tail = None
        else:
            self.tail.next = nd.next
        return nd.data

    def printList(self):
        if self.size > 0:
            ptr = self.tail.next
            while True:
                print(ptr.data, end=" ")
                if ptr == self.tail:
                    break
                ptr = ptr.next
            print("")

    def remove(self, data):
        """
        删除第一个值为 data 的节点
        返回: True 删除成功, False 未找到, None 链表为空
        """
        if self.size == 0:
            return None
        ptr = self.tail
        while ptr.next.data != data:
            ptr = ptr.next
            if ptr == self.tail:
                return False
        self.size -= 1
        if ptr.next == self.tail:
            self.tail = ptr
        ptr.next = ptr.next.next
        return True
```

经典技巧：快慢指针 (Fast-Slow Pointers)
找中点: 快指针走两步，慢指针走一步
判环: 若快慢指针相遇，则链表有环
