```python
import sys
class UnionFind:
    def __init__(self, n):
        self.parent = list(range(n))
        self.rank = [0] * n
        self.num = [1] * n
    
    def find(self, x):
        if self.parent[x]!= x:
            self.parent[x] = self.find(self.parent[x])
        self.num[x]=self.num[self.parent[x]]
        return self.parent[x]
    
    def is_connected(self, x, y):
        return self.find(x) == self.find(y)
    def union(self, x, y):
        root_x = self.find(x)
        root_y = self.find(y)
        if root_x == root_y:
            return
        if self.rank[root_x] < self.rank[root_y]:
            self.parent[root_x] = root_y
            self.num[root_y]+=self.num[root_x]
        elif self.rank[root_x] > self.rank[root_y]:
            self.parent[root_y] = root_x
            self.num[root_x]+=self.num[root_y]
        else:
            self.parent[root_y] = root_x
            self.rank[root_x] += 1
            self.num[root_x]+=self.num[root_y]
ans=0
n,q=map(int,input().split())
uf=UnionFind(n)
for i in range(q):
    u,v=map(int,input().split())
    a,b=uf.find(u-1),uf.find(v-1)
    if a!=b:
        ans+=uf.num[a]*uf.num[b]
        uf.union(a,b)
    print(ans)
import sys

class UnionFind:
    def __init__(self, n):
        self.parent = list(range(n))
        self.rank = [0] * n
        self.size = [1] * n   # 每个根节点的集合大小

    def find(self, x):
        # 迭代路径压缩
        while self.parent[x] != x:
            self.parent[x] = self.parent[self.parent[x]]  # 隔代压缩
            x = self.parent[x]
        return x

    def union(self, x, y):
        """合并两个元素所在的集合，返回两个集合大小的乘积；若已在同一集合，返回0"""
        x_root = self.find(x)
        y_root = self.find(y)
        if x_root == y_root:
            return 0
        # 计算乘积（用合并前的大小）
        prod = self.size[x_root] * self.size[y_root]
        # 按秩合并
        if self.rank[x_root] < self.rank[y_root]:
            self.parent[x_root] = y_root
            self.size[y_root] += self.size[x_root]
        else:
            self.parent[y_root] = x_root
            self.size[x_root] += self.size[y_root]
            if self.rank[x_root] == self.rank[y_root]:
                self.rank[x_root] += 1
        return prod

def main():
    data = sys.stdin.buffer.read().split()
    if not data:
        return
    it = iter(data)
    n = int(next(it))
    q = int(next(it))
    uf = UnionFind(n)
    ans = 0
    out = []
    for _ in range(q):
        u = int(next(it)) - 1
        v = int(next(it)) - 1
        ans += uf.union(u, v)
        out.append(str(ans))
    sys.stdout.write("\n".join(out))

if __name__ == "__main__":
    main()
```
