1. 冒泡排序 (Bubble Sort)
时间复杂度: O(n^2)  空间复杂度: O(1)  稳定排序
原理: 相邻元素比较交换，每趟把最大的"浮"到最后

```python
def bubbleSort(arr):
    n = len(arr)
    for i in range(n):
        swapped = False
        # 最后 i 个元素已经到位
        for j in range(0, n - i - 1):
            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
                swapped = True
        # 如果本趟没有交换，说明已经有序，提前退出
        if not swapped:
            break
```

2. 选择排序 (Selection Sort)
时间复杂度: O(n^2)  空间复杂度: O(1)  不稳定排序
原理: 每趟选出剩余部分的最小(大)值，放到正确位置

```python
def selectionSort(alist):
    for fillslot in range(len(alist) - 1, 0, -1):
        positionOfMax = 0
        for location in range(1, fillslot + 1):
            if alist[location] > alist[positionOfMax]:
                positionOfMax = location
        if positionOfMax != fillslot:
            alist[fillslot], alist[positionOfMax] = alist[positionOfMax], alist[fillslot]
```

3. 插入排序 (Insertion Sort)
时间复杂度: O(n^2), 近乎有序时 O(n)  空间复杂度: O(1)  稳定排序
原理: 构建有序序列，对未排序数据在已排序序列中从后向前扫描并插入

```python
def insertion_sort(arr):
    for i in range(1, len(arr)):
        j = i
        while arr[j - 1] > arr[j] and j > 0:
            arr[j - 1], arr[j] = arr[j], arr[j - 1]
            j -= 1
```

4. 希尔排序 (Shell Sort)
时间复杂度: O(n^2) 最坏  空间复杂度: O(1)  不稳定排序
原理: 插入排序的改进版，通过步长(gap)减少移动次数

```python
def shellSort(arr, n):
    gap = n // 2
    while gap > 0:
        j = gap
        while j < n:
            i = j - gap
            while i >= 0:
                if arr[i + gap] > arr[i]:
                    break
                else:
                    arr[i + gap], arr[i] = arr[i], arr[i + gap]
                i = i - gap
            j += 1
        gap = gap // 2
```

5. 快速排序 (Quick Sort) - 版本1: 双指针法 (pivot取首元素)
时间复杂度: 平均 O(n log n), 最坏 O(n^2)  空间复杂度: O(log n)  不稳定排序
原理: 分治法，选取基准值划分数组，递归排序左右子数组

```python
def quick_sort(arr, low, high):
    """快速排序主函数（递归入口）"""
    if low < high:
        pivot_idx = partition(arr, low, high)
        quick_sort(arr, low, pivot_idx - 1)
        quick_sort(arr, pivot_idx + 1, high)


def partition(arr, low, high):
    """
    分区函数（双指针法 / 双向扫描）
    选取第一个元素作为基准值(pivot)
    left 从左向右找大于pivot的数
    right 从右向左找小于pivot的数
    交换后继续，直到指针相遇，最后将pivot放到相遇位置
    """
    pivot = arr[low]
    left = low + 1
    right = high

    while True:
        while left <= right and arr[left] <= pivot:
            left += 1
        while left <= right and arr[right] >= pivot:
            right -= 1
        if left <= right:
            arr[left], arr[right] = arr[right], arr[left]
        else:
            break

    # 基准值归位: right 指向最后一个 <= pivot 的位置
    arr[low], arr[right] = arr[right], arr[low]
    return right
```

5b. 快速排序 - 版本2: pivot取末元素

```python
def quicksort_v2(arr, left, right):
    if left < right:
        partition_pos = partition_v2(arr, left, right)
        quicksort_v2(arr, left, partition_pos - 1)
        quicksort_v2(arr, partition_pos + 1, right)


def partition_v2(arr, left, right):
    i = left
    j = right - 1
    pivot = arr[right]
    while i <= j:
        while i <= right and arr[i] < pivot:
            i += 1
        while j >= left and arr[j] >= pivot:
            j -= 1
        if i < j:
            arr[i], arr[j] = arr[j], arr[i]
    if arr[i] > pivot:
        arr[i], arr[right] = arr[right], arr[i]
    return i
```

6. 归并排序 (Merge Sort) - 递归版 (自上而下)
时间复杂度: O(n log n) 所有情况  空间复杂度: O(n)  稳定排序
原理: 分治法，将数组不断二分直到长度为1，然后两两合并

```python
def merge_sort(arr):
    if len(arr) <= 1:
        return arr
    mid = len(arr) // 2
    left = merge_sort(arr[:mid])
    right = merge_sort(arr[mid:])
    return merge(left, right)


def merge(left, right):
    result = []
    i = j = 0
    while i < len(left) and j < len(right):
        if left[i] <= right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1
    result.extend(left[i:])
    result.extend(right[j:])
    return result
```

6b. 归并排序 - 原地版 (自上而下)

```python
def mergeSort_inplace(arr):
    if len(arr) > 1:
        mid = len(arr) // 2
        L = arr[:mid]
        R = arr[mid:]
        mergeSort_inplace(L)
        mergeSort_inplace(R)
        i = j = k = 0
        while i < len(L) and j < len(R):
            if L[i] <= R[j]:
                arr[k] = L[i]
                i += 1
            else:
                arr[k] = R[j]
                j += 1
            k += 1
        while i < len(L):
            arr[k] = L[i]
            i += 1
            k += 1
        while j < len(R):
            arr[k] = R[j]
            j += 1
            k += 1
```

6c. 归并排序 - 迭代版 (自下而上)

```python
def merge_sort_iterative(arr):
    n = len(arr)
    size = 1
    while size < n:
        for start in range(0, n, 2 * size):
            mid = min(start + size, n)
            end = min(start + 2 * size, n)
            arr[start:end] = merge(arr[start:mid], arr[mid:end])
        size *= 2
    return arr
```

排序算法对比表
算法        | 平均时间  | 最坏时间  | 空间   | 稳定性
------------|-----------|-----------|--------|-------
Bubble      | O(n^2)    | O(n^2)    | O(1)   | Stable
Selection   | O(n^2)    | O(n^2)    | O(1)   | Unstable
Insertion   | O(n^2)    | O(n^2)    | O(1)   | Stable
Shell       | O(n^1.3)  | O(n^2)    | O(1)   | Unstable
Quick       | O(nlogn)  | O(n^2)    | O(logn)| Unstable
Merge       | O(nlogn)  | O(nlogn)  | O(n)   | Stable

测试代码

```python
if __name__ == "__main__":
    import copy

    original = [64, 34, 25, 12, 22, 11, 90]

    # 冒泡排序
    arr = copy.copy(original)
    bubbleSort(arr)
    print(f"冒泡排序: {arr}")

    # 选择排序
    arr = copy.copy(original)
    selectionSort(arr)
    print(f"选择排序: {arr}")

    # 插入排序
    arr = copy.copy(original)
    insertion_sort(arr)
    print(f"插入排序: {arr}")

    # 希尔排序
    arr = copy.copy(original)
    shellSort(arr, len(arr))
    print(f"希尔排序: {arr}")

    # 快速排序
    arr = copy.copy(original)
    quick_sort(arr, 0, len(arr) - 1)
    print(f"快速排序: {arr}")

    # 归并排序 (返回新数组)
    arr = merge_sort(copy.copy(original))
    print(f"归并排序: {arr}")

    # 归并排序 (原地)
    arr = copy.copy(original)
    mergeSort_inplace(arr)
    print(f"归并排序(原地): {arr}")

    # 归并排序 (迭代)
    arr = copy.copy(original)
    merge_sort_iterative(arr)
    print(f"归并排序(迭代): {arr}")
```
